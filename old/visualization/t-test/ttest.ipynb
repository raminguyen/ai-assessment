## Pair T-test for whole dataset

```python
import pandas as pd
from scipy import stats

df = pd.read_csv("dataset_full.csv")

#df[["E1 Gen", "E2 Gen"]].describe()

result = stats.ttest_rel(df['E1 Gen'], df['E2 Gen'])

print("Paired t-test Results:")
print("  t-statistic: " + str(round(result.statistic, 2)))
print("  p-value: " + str(round(result.pvalue, 3)))
print("  degrees of freedom: " + str(result.df))
```
